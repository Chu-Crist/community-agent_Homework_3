```python
import pandas as pd
from tools.data import load_data
from tools.graph import build_graph
from tools.communities import detect_communities, community_sizes, top_users_by_centrality
from tools.viz import plot_network

# 1. Load the REAL data from the csv file
# We use a high limit to get as much data as possible
df = pd.read_csv('data/sample.csv')
print(f"Loaded {len(df)} rows of real data.")

# 2. Build the graph
graph = build_graph(df)
print(f"Graph has {graph.number_of_nodes()} users and {graph.number_of_edges()} connections.")

# 3. Detect communities
communities = detect_communities(graph)
sizes = community_sizes(communities)
top_users = top_users_by_centrality(graph, communities)

# 4. Print the real results
print("\n=== REAL RESULTS ===")
print("Total Communities detected:", len(sizes))
print("Top 5 community sizes:", list(sizes.items())[:5])
print("Top 2 users in the largest community:", top_users.get(list(sizes.keys())[0], [])[:2])

# 5. Generate the real plot
img_path = plot_network(graph, communities, out_path="outputs/real_network.png")
print("\nReal image saved to:", img_path)
```
